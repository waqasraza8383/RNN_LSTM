```python
docs=[
    'go india',
    'india india',
    'hip hip hurray',
    'jeetga bhai jeetega india jeetega',
    'bharat mata ki jai',
    'kohli kohli',
    'sachin sachin',
    'dhoni dhoni',
    'modi ji ji jai',
    'inquilab zindabad'
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer.fit_on_texts(docs)
```
